## Purpose

The purpose of this article was to show how music is connected. Their prime example was how "Hit Em Up" by Tupac was (through a list of connections) inspired by "Hall of the Mountain King" a classical song that primarily instrumental. 



But it's not just that a song from a vastly different genre and time period can be used as inspiration for another song, but also songs have many inspirations making for even more connections, and that two very songs can have the same song inspired it to be made.

## Data and Evidence

As stated before they found connections in one of four ways. Samples, Covers, Interpolations, or Remixing. They used Genius.com to figure out the basic connections (Genius has an API that shows how songs are connected). After this they cross referenced with other sites, WhoSampled.com, Spotify, news media, and academic scholarship. These other sites allowed them to revise and redefine the definitions of their connections within their dataset. Connections were determined by a mix of author interpretation, credits from artists, third party interpretation, and sonic evidence. One challenge that was acknowledged with their article was that there are legal implications with sampling and in some instances samples are legally acknowledged and in other instances they are not. Because of this there is a strong possibility that there are even more connections between songs that are unable to be acknowledged due to legal reasons.

Navigation is where I feel this project can fall short. It's very clearly meant to be navigatated via the mouse's scroll wheel, and if you try to navigate using your keys some parts get skipped over and don't present properly. Another are of critisim I have for this project is that the web maps aren't interactable. I think allowing users to interact with the web maps would be something that would make this project more engageing and could further show how songs the are connected.
